# DB（SQLite）を使う時の処理の流れ

1.DBへのコネクションを確立

2.コネクションオブジェクトからデータ操作用のカーソルオブジェクトを取得

3.実行したいSQLを定義

4.SQLを実行

5.必要があればコミットする

6.コネクションを閉じる

```python
path = ''
db_name = 'car.db'

try:
    # DB接続オブジェクトの作成
    conn = sqlite3.connect(path + db_name)

    # カーソルオブジェクトの作成
    cur = conn.cursor()

    # SQL文の作成
    # テーブルの作成
    sql = 'CREATE TABLE car (id INT, name TEXT, price REAL);'

    # SQL文の実行
    cur.execute(sql)
except sqlite3.Error as e:
    print('エラーが発生しました:', e)
finally:
    # DBへの接続を閉じる
    conn.close()
```

## データをDBに挿入する（単一レコード）


```python
path = ''
db_name = 'car.db'
# DB接続オブジェクトの作成
conn = sqlite3.connect(path + db_name)

# カーソルオブジェクトの作成
cur = conn.cursor()

# SQL文の作成
# データの挿入
sql = "INSERT INTO car (id, name, price) VALUES (1, '370z', 7000000);"

# SQL文の実行
cur.execute(sql)

# 変更をDBに反映させる
conn.commit()

# DBへの接続を閉じる
conn.close()
```

### データをDBに挿入する（複数レコード）

```python
path = ''
db_name = 'car.db'
# DB接続オブジェクトの作成
conn = sqlite3.connect(path + db_name)

# カーソルオブジェクトの作成
cur = conn.cursor()

# SQL文の作成
# 複数レコードの挿入
sql = "INSERT INTO car (id, name, price) VALUES (?, ?, ?);"

cars = [
    (2, 'Mustang', 5500000),
    (3, 'Q3', 8300000),
    (4, 'X1', 8000000)
]
# SQL文の実行
cur.executemany(sql, cars)

# 変更をDBに反映させる
conn.commit()

# DBへの接続を閉じる
conn.close()
```

### DB内のデータを参照する

```python
path = ''
db_name = 'car.db'
try:
    # DB接続オブジェクトの作成
    conn = sqlite3.connect(path + db_name)

    # カーソルオブジェクトの作成
    cur = conn.cursor()

    # SQL文の作成
    # データの挿入
    sql = "SELECT * FROM car;"

    # SQL文の実行
    cur.execute(sql)

    # 変更をDBに反映させる
    # conn.commit()
except sqlite3.Error as e:
    print('エラーが発生しました:', e)
else:
    for row in cur:
        id, name, price = row
        print(id, name, price)
finally:
    # DBへの接続を閉じる
    conn.close()
```

### データの削除

```python
path = ''
db_name = 'car.db'
try:
    # DB接続オブジェクトの作成
    conn = sqlite3.connect(path + db_name)

    # カーソルオブジェクトの作成
    cur = conn.cursor()

    # SQL文の作成
    # データの削除
    sql = "DELETE FROM car WHERE id = 1;"

    # SQL文の実行
    cur.execute(sql)

    # 変更をDBに反映させる
    conn.commit()
except sqlite3.Error as e:
    print('エラーが発生しました:', e)
else:
    for row in cur:
        id, name, price = row
        print(id, name, price)
finally:
    # DBへの接続を閉じる
    conn.close()
```

##  データの更新

```python
path = ''
db_name = 'car.db'
try:
    # DB接続オブジェクトの作成
    conn = sqlite3.connect(path + db_name)

    # カーソルオブジェクトの作成
    cur = conn.cursor()

    # SQL文の作成
    # データの更新
    sql_update = "UPDATE car SET price = ? WHERE id = ?;"

    # SQL文の実行
    cur.execute(sql_update,(5000000, 2))

    # 変更をDBに反映させる
    conn.commit()
except sqlite3.Error as e:
    print('エラーが発生しました:', e)
else:
    for row in cur:
        id, name, price = row
        print(id, name, price)
finally:
    # DBへの接続を閉じる
    conn.close()
```
